**Python OOPs Questions**

1. What is Object-Oriented Programming (OOP)

   => Object-Oriented Programming is a programming approach where the code is organized into objects instead of functions. These objects represent real-world things and contain both data and the functions that work on that data. OOP makes programs easier to understand, maintain, and reuse.


2. What is a class in OOP

   => A class is a blueprint used to create objects. It defines what data an object will hold and what actions it can perform. Without a class, objects cannot be created.


3. What is an object in OOP

   => An object is a real instance of a class. It represents something that actually exists in memory and can access the properties and methods defined in its class.


4. What is the difference between abstraction and encapsulation.
  
   => Abstraction means showing only the important features of an object and hiding the internal details. Encapsulation means binding data and methods together in a class and protecting the data from direct access.


5. What are dunder methods in Python

   => Dunder methods are special methods in Python that begin and end with double underscores. They are used to define how objects behave, such as object creation, printing, or using operators.


6. Explain the concept of inheritance in OOP

   => Inheritance is a feature in OOP where one class can acquire the properties and methods of another class. It helps in reusing existing code and makes programs easier to extend and modify.


7. What is polymorphism in OOP

   => Polymorphism means "many forms." It allows the same method or function name to behave differently depending on the object that is using it. This makes programs more flexible and easier to understand.


8. How is encapsulation achieved in Python ?

   => Encapsulation in Python is achieved by defining variables and methods inside a class and controlling access to them. Python uses naming conventions like single or double underscores to indicate protected and private data.


9. What is a constructor in Python

   => A constructor is a special method that runs automatically when an object is created. In Python, it is usually used to initialize the object's data.


10. What are class and static methods in Python

    => Class methods work with the class itself and can access class variables. Static methods do not use class or object data and are mainly used for utility or helper tasks.


11. What is method overloading in Python

    => Method overloading means using the same method name with different parameters. In Python, it is usually handled by using default or variable arguments instead of creating multiple methods.


12. What is method overriding in OOP

    => Method overriding happens when a child class provides its own version of a method that is already defined in the parent class. This allows the child class to change or improve the behavior of that method.


13. What is a property decorator in Python

    => A property decorator in Python is used to access a method like an attribute. It helps control how a value is read or updated while keeping the code clean and easy to use.


14. Why is polymorphism important in OOP

    => Polymorphism is important because it allows the same code to work with different types of objects. This reduces complexity, improves flexibility, and makes programs easier to maintain.


15. What is an abstract class in Python

    => An abstract class is a class that cannot be used to create objects directly. It is meant to be inherited and ensures that child classes implement certain methods.


16. What are the advantages of OOP

    => OOP makes programs easier to understand and organize. It helps in code reuse, improves security through data hiding, and makes applications easier to update and maintain in the future.


17. What is the difference between a class variable and an instance variable
    => A class variable is shared by all objects of a class, so the same value is used everywhere. An instance variable belongs to a specific object, and each object can have its own separate value.


18. What is multiple inheritance in Python
    
    => Multiple inheritance is a feature where a class can inherit properties and methods from more than one parent class. This allows a class to combine behaviors from different classes.

19. Explain the purpose of '__str__' and '__repr__'  methods in Python

    => The __str__ method is used to return a readable and user-friendly description of an object.
    The __repr__ method is used to return a detailed and technical representation of the object, mainly for developers and debugging.


20. What is the significance of the 'super()' function in Python
  
    => The super() function is used to call methods of a parent class from a child class. It helps reuse parent class code and supports clean and proper inheritance.


21. What is the significance of the __del__ method in Python

    => The __del__ method is called when an object is about to be destroyed. It is mainly used to release resources like files or memory before the object is removed.


22. What is the difference between @staticmethod and @classmethod in Python ?

    => A static method does not use class or object data and works like a normal function inside a class. A class method uses the class itself as a parameter and can access or modify class variables.


23. How does polymorphism work in Python with inheritance

24. What is method chaining in Python OOP

    => Method chaining is a technique where multiple methods are called one after another on the same object in a single line. This works when each method returns the object itself, making the code shorter and easier to read.


25. What is the purpose of the __call__ method in Python?

```python
# 1.Create a parent class Animal with a method speak() that prints a generic message.
    #Create a child class Dog that overrides the speak() method to print "Bark!".

class Animal:
    def speak(self):
        print("Animal sound")

class Dog(Animal):
    def speak(self):
        print("Bark!")

d = Dog()
d.speak()
```

```python
# 2. Write a program to create an abstract class Shape with a method area().
    #Derive classes Circle and Rectangle from it and implement the area() method in both.

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def area(self):
        return 3.14 * 5 * 5

class Rectangle(Shape):
    def area(self):
        return 4 * 6

c = Circle()
r = Rectangle()

print("Circle area:", c.area())
print("Rectangle area:", r.area())
```

```python
# 3. Implement a multi-level inheritance scenario where a class Vehicle has an attribute type.
    #Derive a class Car and further derive a class ElectricCar that adds a battery attribute.

class Vehicle:
    def __init__(self, t):
        self.type = t

class Car(Vehicle):
    pass

class ElectricCar(Car):
    def __init__(self, t, battery):
        Vehicle.__init__(self, t)
        self.battery = battery

e = ElectricCar("Car", "5000mAh")
print(e.type)
print(e.battery)
```

```python
# 4. Demonstrate polymorphism by creating a base class Bird with a method fly().
    #Create two derived classes Sparrow and Penguin that override the fly() method.

class Bird:
    def fly(self):
        print("Bird can fly")

class Sparrow(Bird):
    def fly(self):
        print("Sparrow flies")

class Penguin(Bird):
    def fly(self):
        print("Penguin cannot fly")

b1 = Sparrow()
b2 = Penguin()

b1.fly()
b2.fly()
```

```python
# 5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes balance and methods to deposit, withdraw, and check balance

class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amt):
        self.__balance += amt

    def withdraw(self, amt):
        self.__balance -= amt

    def show_balance(self):
        print("Balance:", self.__balance)

acc = BankAccount()
acc.deposit(1000)
acc.withdraw(300)
acc.show_balance()
```

```python
# 6. Demonstrate runtime polymorphism using a method play() in a base class Instrument.
    #Derive classes Guitar and Piano that implement their own version of play().

class Instrument:
    def play(self):
        print("Instrument plays sound")

class Guitar(Instrument):
    def play(self):
        print("Guitar is playing")

class Piano(Instrument):
    def play(self):
        print("Piano is playing")

i1 = Guitar()
i2 = Piano()

i1.play()
i2.play()
```

```python
# 7. Create a class MathOperations with a class method add_numbers()
    #to add two numbers and a static method subtract_numbers() to subtract two numbers.

class MathOperations:

    @classmethod
    def add_numbers(cls, a, b):
        return a + b

    @staticmethod
    def subtract_numbers(a, b):
        return a - b

print(MathOperations.add_numbers(5, 3))
print(MathOperations.subtract_numbers(5, 3))
```

```python
# 8. Implement a class Person with a class method to count the total number of persons created.

class Person:
    count = 0

    def __init__(self):
        Person.count += 1

    @classmethod
    def total_persons(cls):
        print("Total persons:", cls.count)

p1 = Person()
p2 = Person()

Person.total_persons()
```

```python
# 9. Write a class Fraction with attributes numerator and denominator.
    #Override the str method to display the fraction as "numerator/denominator".

class Fraction:
    def __init__(self, n, d):
        self.numerator = n
        self.denominator = d

    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

f = Fraction(2, 5)
print(f)
```

```python
# 10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two vectors.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def show(self):
        print(self.x, self.y)

v1 = Vector(2, 3)
v2 = Vector(4, 5)

v3 = v1 + v2
v3.show()
```

```python
# 11. Create a class Person with attributes name and age.
      #Add a method greet() that prints "Hello, my name is {name} and I am {age} years old."

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

p = Person("Mayur", 20)
p.greet()
```

```python
# 12. Implement a class Student with attributes name and grades.
    #Create a method average_grade() to compute the average of the grades.

class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def average_grade(self):
        return sum(self.grades) / len(self.grades)

s = Student("Anita", [80, 75, 90])
print(s.average_grade())
```

```python
# 13. Create a class Rectangle with methods set_dimensions() to set the dimensions and area() to calculate the area.

class Rectangle:
    def set_dimensions(self, l, b):
        self.length = l
        self.breadth = b

    def area(self):
        return self.length * self.breadth

r = Rectangle()
r.set_dimensions(4, 5)
print(r.area())
```

```python
# 14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked and hourly rate. Create a derived class Manager that adds a bonus to the salary.

class Employee:
    def calculate_salary(self, hours, rate):
        return hours * rate

class Manager(Employee):
    def calculate_salary(self, hours, rate, bonus):
        return super().calculate_salary(hours, rate) + bonus

m = Manager()
print(m.calculate_salary(40, 100, 500))
```

```python
# 15. Create a class Product with attributes name, price, and quantity.
    #Implement a method total_price() that calculates the total price of the product.

class Product:
    def __init__(self, name, price, qty):
        self.name = name
        self.price = price
        self.quantity = qty

    def total_price(self):
        return self.price * self.quantity

p = Product("Pen", 10, 5)
print(p.total_price())
```

```python
# 16. Create a class Animal with an abstract method sound().
     #Create two derived classes Cow and Sheep that implement the sound() method.

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Cow(Animal):
    def sound(self):
        print("Moo")

class Sheep(Animal):
    def sound(self):
        print("Baa")

c = Cow()
s = Sheep()
c.sound()
s.sound()
```

```python
# 17. Create a class Book with attributes title, author, and year_published.
     #Add a method get_book_info() that returns a formatted string with the book's details.

class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year_published = year

    def get_book_info(self):
        return f"{self.title} by {self.author} ({self.year_published})"

b = Book("Python Basics", "John", 2022)
print(b.get_book_info())
```

```python
# 18. Create a class House with attributes address and price.
     #Create a derived class Mansion that adds an attribute number_of_rooms.

class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

class Mansion(House):
    def __init__(self, address, price, rooms):
        super().__init__(address, price)
        self.number_of_rooms = rooms

m = Mansion("Mumbai", 5000000, 10)
print(m.address, m.number_of_rooms)
```
